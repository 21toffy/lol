```python
import os
import torch
import soundfile as sf
from tqdm import tqdm
from pydub import AudioSegment
import re
from google.colab import files
from IPython.display import Audio, display
import numpy as np
from qwen_tts import Qwen3TTSModel

# ========================= CONFIG =========================
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
DTYPE = torch.bfloat16 # Changed from torch.float16 for better stability

# Use CustomVoice model for preset speakers like Ryan
BASE_MODEL = "Qwen/Qwen3-TTS-12Hz-0.6B-CustomVoice"   # or 1.7B version if you have VRAM

OUTPUT_DIR = "tts_segments"
FINAL_OUTPUT = "full_narration.wav"
STATE_FILE = "done_segments.txt"

# ====================== BEEP FUNCTION ======================
def play_beep():
    sample_rate = 22050
    duration = 0.35
    frequency = 950
    t = np.linspace(0, duration, int(sample_rate * duration))
    wave = 0.4 * np.sin(2 * np.pi * frequency * t)
    display(Audio(wave, rate=sample_rate, autoplay=True))

# ====================== SETUP ======================
torch.cuda.empty_cache()
os.makedirs(OUTPUT_DIR, exist_ok=True)

model = Qwen3TTSModel.from_pretrained(
    BASE_MODEL,
    device_map=DEVICE,
    torch_dtype=DTYPE,
    attn_implementation="sdpa"
)
print(f"✅ Model loaded: {BASE_MODEL} on {DEVICE}")

# ====================== INSTRUCTION FOR RYAN ======================
INSTRUCTION = """Speak very slowly with a deep, confident, dramatic tone.
Use long pauses between sentences. Mysterious and deliberate delivery.
Take your time."""

# ====================== SPLIT TEXT ======================
def split_text_by_double_newlines(text):
    chunks = re.split(r'\n\s*\n\s*\n', text.strip())
    return [chunk.strip() for chunk in chunks if chunk.strip()]

# ====================== YOUR TEXT (unchanged) ======================
full_text = text   # ← Your variable stays exactly as you have it

chunks = split_text_by_double_newlines(full_text)
print(f"Split into {len(chunks)} chunks.")

# ====================== PROGRESS ======================
done = set()
if os.path.exists(STATE_FILE):
    with open(STATE_FILE, 'r') as f:
        done = {line.strip() for line in f if line.strip()}

print(f"Already done: {len(done)} / {len(chunks)}")

# ====================== GENERATION ======================
for i, chunk in enumerate(tqdm(chunks, desc="Generating audio")):
    chunk_id = f"chunk_{i:04d}"
    output_path = os.path.join(OUTPUT_DIR, f"{chunk_id}.wav")

    if chunk_id in done or os.path.exists(output_path):
        continue

    try:
        torch.cuda.empty_cache()

        wavs, sr = model.generate_custom_voice(
            text=chunk,
            language="English",
            speaker="Ryan",           # Preset voice
            instruct=INSTRUCTION      # Style control (not read aloud)
        )

        audio = wavs[0] if isinstance(wavs, (list, tuple)) else wavs
        sf.write(output_path, audio, sr)

        with open(STATE_FILE, 'a') as f:
            f.write(chunk_id + "\n")

        print(f"✅ Chunk {i+1}/{len(chunks)} completed")
        play_beep()

    except Exception as e:
        print(f"\nError on chunk {i}: {e}")
        torch.cuda.empty_cache()

print("\n🎉 All chunks finished!")

# ====================== MERGE ======================
def merge_audio():
    segments = sorted([f for f in os.listdir(OUTPUT_DIR) if f.endswith('.wav')])
    if not segments:
        print("No segments found!")
        return

    combined = AudioSegment.from_wav(os.path.join(OUTPUT_DIR, segments[0]))
    for seg in tqdm(segments[1:], desc="Merging"):
        combined += AudioSegment.from_wav(os.path.join(OUTPUT_DIR, seg))

    combined.export(FINAL_OUTPUT, format="wav")
    print(f"✅ Final file saved: {FINAL_OUTPUT}")
    files.download(FINAL_OUTPUT)

merge_audio()
```
